# eegnet

## scaling

```python
plt.plot([0.5, 1.5, 2, 2.5],[63.4, 63.5, 63.4, 63.8])
plt.plot([0.5, 1.5, 2, 2.5], [61.7, 60, 56.9, 51.3])
plt.plot([0.5, 1.5, 2, 2.5], [57.5, 62.5, 59.9, 43.3])
plt.ylim(40,70)
plt.legend(['Both', 'Train', 'Test'])
```

## noise
