# verify the dataset loading

```python
import sys
import os

sys.path.insert(0, os.path.abspath("../src"))

from eeg_robustness.data import load_dataset
```

```python
from collections import Counter

import numpy as np

ds = load_dataset()

print(
    f"Trials={ds.n_trials}, "
    f"Subjects={len(np.unique(ds.subjects))}, "
    f"Classes={ds.n_classes}, "
    f"Channels={ds.n_channels}, "
    f"Samples={ds.n_samples}, "
    f"Fs={ds.sampling_rate}"
)

print(f"Original labels: {ds.label_values.tolist()}")

if ds.sessions is not None:
    print(f"Sessions: {np.unique(ds.sessions).tolist()}")

if ds.runs is not None:
    print(f"Runs: {np.unique(ds.runs).tolist()}")

print()

for subject in np.unique(ds.subjects):
    mask = ds.subjects == subject
    counts = Counter(ds.y[mask].tolist())

    print(
        f"S{subject}: {mask.sum()} trials | "
        + ", ".join(
            f"class {i}: {counts.get(i, 0)}"
            for i in range(ds.n_classes)
        )
    )

assert ds.n_trials == len(ds.y) == len(ds.subjects)
assert np.isfinite(ds.X).all()
assert ds.X.ndim == 3
assert ds.n_classes == 4
assert len(np.unique(ds.subjects)) == 9

if ds.sessions is not None:
    assert len(ds.sessions) == ds.n_trials

if ds.runs is not None:
    assert len(ds.runs) == ds.n_trials

print("\nAll basic checks passed.")
```
